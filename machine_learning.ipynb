```python
import csv
a = []
with open('enjoysport.csv', 'r') as csvfile:
    for row in csv.reader(csvfile):
        a.append(row)
    print(a)
print("\n The total number of training instances are : ",len(a))
num_attribute = len(a[0])-1
print("\n The initial hypothesis is : ")
hypothesis = ['0']*num_attribute
print(hypothesis)
for i in range(0, len(a)):
 if a[i][num_attribute] == 'yes':
  for j in range(0, num_attribute):
      if hypothesis[j] == '0' or hypothesis[j] == a[i][j]:
          hypothesis[j] = a[i][j]
      else:
          hypothesis[j] = '?'
 print("\n The hypothesis for the training instance {} is :\n" .format(i+1),hypothesis)
print("\n The Maximally specific hypothesis for the training instance is ")
print(hypothesis)
```

```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv("iris_dataset.csv")

print("IRIS DATASET:")
print(df.head())

# Entropy
def entropy(y):
    _, counts = np.unique(y, return_counts=True)
    p = counts / len(y)
    return -sum(p * np.log2(p))

# Information Gain
def gain(df, feature, threshold):
    left = df[df[feature] <= threshold]
    right = df[df[feature] > threshold]

    if len(left) == 0 or len(right) == 0:
        return 0

    return entropy(df["class"]) - (
        len(left) / len(df) * entropy(left["class"]) +
        len(right) / len(df) * entropy(right["class"])
    )

# Find best ID3 split
features = [
    "sepal_length_cm",
    "sepal_width_cm",
    "petal_length_cm",
    "petal_width_cm"
]

best_gain = 0
best_feature = ""
best_threshold = 0

for feature in features:
    values = sorted(df[feature].unique())

    for i in range(len(values) - 1):
        threshold = (values[i] + values[i + 1]) / 2
        g = gain(df, feature, threshold)

        if g > best_gain:
            best_gain = g
            best_feature = feature
            best_threshold = threshold

print("\nID3 BEST SPLIT:")
print("Feature:", best_feature)
print("Threshold:", round(best_threshold, 2))
print("Information Gain:", round(best_gain, 4))

# New sample
sample = {
    "sepal_length_cm": 5.9,
    "sepal_width_cm": 3.0,
    "petal_length_cm": 5.1,
    "petal_width_cm": 1.8
}

# Classification
if sample[best_feature] <= best_threshold:
    prediction = "setosa"
else:
    prediction = "not setosa"

print("\nNEW SAMPLE:")
print(sample)

print("\nPREDICTED CLASS:", prediction)
```
